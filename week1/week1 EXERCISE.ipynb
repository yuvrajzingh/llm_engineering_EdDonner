```python
# imports
import os
from dotenv import load_dotenv
from openai import OpenAI
```

```python
# constants

MODEL_GPT = 'openai/gpt-4o'
MODEL_LLAMA = 'llama3.2'
BASE_URL = "https://models.github.ai/inference"
```

```python
# set up environment
load_dotenv(override=True)
api_key = os.getenv('OPENAI_API_KEY')
openai = OpenAI(base_url=BASE_URL, api_key=api_key)

if api_key and len(api_key)>10:
    print("API key looks good so far")
else:
    print("There might be a problem with your API key? Please visit the troubleshooting notebook!")
```

```python
# here is the question; type over this to ask something new

question = """
Please explain what this code does and why:
yield from {book.get("author") for book in books if book.get("author")}
"""
```

```python
# Get gpt-4o-mini to answer, with streaming
def techAssistant(question):
    system_prompt = """
    You are a snarky assistant that analyzes the technical question or statement asked by the user and gives a short, snarky, humorous answer. 
    """
    messages = [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": question}
    ]
    response = openai.chat.completions.create(model=MODEL_GPT, messages=messages)
    return response.choices[0].message.content

print(techAssistant(question))
    
    
```
